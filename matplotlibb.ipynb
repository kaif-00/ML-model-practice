# Matplotlib Lessions

```python
#pyplot have all the function to draw graphs
import matplotlib.pyplot as plt

x = [1,2,3,4,9]
y = [22,44,12,56,90]

plt.plot(x,y)
plt.show()
```

```python
x = ['mon','tue','wed','thr','fri','sat']
y = [22,44,62,36,90,12]

plt.plot(x,y,marker='o', label = 'Sale')
plt.xlabel('days')
plt.ylabel('sales')
plt.title('Sales Data')
plt.legend(loc='upper left')
# plt xlim, ylim increase the value of the axes or decrease according to value
# plt.ylim(0,2000)
plt.grid()
plt.show()
```

```python
# Bar
x = ['Gorakhpur','Lucknow','Mumbai','Banglor']
y = [23000,45000,91200,120004]

plt.bar(x,y)
plt.xlabel('city')
plt.ylabel('price')
plt.title('house price Data')
# plt.grid()

plt.show()
```

```python
# Piechart
x = ['Gorakhpur','Lucknow','Mumbai','Banglor']
y = [23000,45000,91200,120004]

#pie(y,labels=x,autopct='%1.1f%%') autopct generate percentage


plt.pie(y,labels=x,autopct='%1.1f%%')
plt.xlabel('city')
plt.ylabel('price')
plt.title('house price Data')
plt.grid()

plt.show()
```

```python
#histogram

score=[45,67,45,67,48,39,67,64,34,57,76,76,44,54,86,84,90,95,65,67,98]

plt.hist(score,bins=10,edgecolor='black')

plt.xlabel('score')
plt.title('Exam score')
plt.show()

#so on y axis it shows where data is concentrated by showing how many value are present in particular range
# like 2 students have scored less then 40 
```

```python
# Scatter plot represent points of data on a graph with two values x,y

x =[1,2,3,4,5,6,7,8]
y = [34,45,56,72,79,84,89,92]

plt.scatter(x,y,color='red',marker='o',label = 'Student Data')
plt.xlabel('Houres')
plt.ylabel('Marks')
plt.title('Exam score')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Subplot is used to make multiple graphs at the same time

x =[1,2,3,4,5,6,7,8]
y = [34,45,56,72,79,84,89,92]

plt.subplot(1,2,1)# row, column, index
plt.scatter(x,y,color='red',marker='o',label = 'Student Data')
plt.xlabel('Houres')
plt.ylabel('Marks')
plt.title('Exam score')
plt.grid(True)
plt.legend()


plt.subplot(1,2,2)
plt.bar(x,y,label = 'Student Data')
plt.xlabel('Houres')
plt.ylabel('Marks')
plt.title('Exam score')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Subplot is used to make multiple graphs at the same time

x =[1,2,3,4,5,6,7,8]
y = [34,45,56,72,79,84,89,92]

fig, ax = plt.subplots(1,2,figsize=(10,4))# row, column, figure size
ax[0].scatter(x,y,color='red',marker='o',label = 'Student Data')
ax[0].set_xlabel('Houres')
ax[0].set_ylabel('Marks')
ax[0].set_title('Exam score')
ax[0].grid(True)
ax[0].legend()


ax[1].bar(x,y,label = 'Student Data')
ax[1].set_xlabel('Houres')
ax[1].set_ylabel('Marks')
ax[1].set_title('Exam score')
ax[1].grid(True)
ax[1].legend()

fig.suptitle('Comparing two graphs')
plt.tight_layout()
plt.show()
```

```python
# Save figure

x =[1,2,3,4,5,6,7,8]
y = [34,45,56,72,79,84,89,92]

plt.scatter(x,y,color='red',marker='o',label = 'Student Data')
plt.xlabel('Houres')
plt.ylabel('Marks')
plt.title('Exam score')
plt.grid(True)
plt.legend()

plt.savefig('data.png',dpi=300,bbox_inches='tight')
plt.show()
```
