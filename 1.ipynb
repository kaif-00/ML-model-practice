```python
from sklearn.linear_model import LinearRegression
import pandas as pd

# Linear Regression is a supervised machine learning algorithm that is used to predict a continuous target variable based on one or more input features. It assumes a linear relationship between the input features and the target variable.

data = {
    'Study_hour':[1,2,3,4,5],
    'Exam_Score':[33,45,57,64,75]
}
df = pd.DataFrame(data)
model = LinearRegression()
model.fit(df[['Study_hour']],df['Exam_Score'])
```

```python
from sklearn.linear_model import LogisticRegression
import pandas as pd

# Logistic Regression is a supervised machine learning algorithm that is used to predict a binary target variable (0 or 1) based on one or more input features. It estimates the probability of an event occurring based on the input features and uses a logistic function to model the relationship between the input features and the target variable.

data = { "Study_hour":[1,2,3,4,5],
         "Pass":[0,0,1,1,1]}

df = pd.DataFrame(data)
model = LogisticRegression()

model.fit(df[['Study_hour']], df['Pass'])
hours = float(input())
f = model.predict([[hours]])[0]

if f == 1:
    print("You will pass the exam")
else:
    print("You will fail the exam")
```

```python
from sklearn.neighbors import KNeighborsClassifier

# K-Nearest Neighbors (KNN) is a supervised machine learning algorithm that is used for classification only outputs (0,1) and regression tasks. It works by finding the k-nearest data points to a given input and making predictions based on the majority class (for classification) or the average value (for regression) of those neighbors.

data = { "Study_hour":[1,2,3,4,5],
         "Pass":[0,0,1,1,1]}

df = pd.DataFrame(data)

model = KNeighborsClassifier(n_neighbors=3)

model.fit(df[['Study_hour']], df['Pass'])

hours = float(input())

f = model.predict([[hours]])[0]
f
```

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score,confusion_matrix

y_true = [1,0,1,0,0,1,0]
y_pred = [0,0,0,1,0,1,0]

#accuracy is the ratio of correctly predicted observations to the total observations. It is a good measure when the target classes are well balanced.
accuracy = accuracy_score(y_true, y_pred)
print("Accuracy:", accuracy)

#precision is the ratio of correctly predicted positive observations to the total predicted positive observations. It is a good measure when the cost of false positives is high.
precision = precision_score(y_true, y_pred)
print('Precision:', precision)

#recall is the ratio of correctly predicted positive observations to the total actual positive observations. It is a good measure when the cost of false negatives is high.
recall = recall_score(y_true, y_pred)
print('Recall:', recall)

#f1 score is the weighted average of precision and recall. It is a good measure when you need to balance precision and recall.
f1 = f1_score(y_true, y_pred)
print('F1 score:', f1)

confusion_matrix = confusion_matrix(y_true, y_pred)
print('Confusion Matrix:',confusion_matrix)
```

```python
# MAE MSE RMSE
from sklearn.metrics import mean_absolute_error, mean_squared_error, root_mean_squared_error

real_score = [30,40,70,50,90,100]
predicted_score = [30,40,70,90,50,100]

#MAE (Mean Absolute Error) is the average of the absolute differences between the predicted and actual values. It gives an idea of how wrong the predictions were.
mae = mean_absolute_error(real_score, predicted_score)

#MSE (Mean Square Error) is average of squared values different between the predicted and actual values. it gives large error more weight than smaller errors. It is useful when large errors are particularly undesirable.
mse = mean_squared_error(real_score, predicted_score)

#RMSE (Root Mean Square Error) is the square root of the average of squared differences between prediction and actual observation. It is useful when large errors are particularly undesirable.
rmse = root_mean_squared_error(real_score, predicted_score)

print("MAE:", mae)
print("MSE:", mse)
print("RMSE:", rmse)
```
